```python
num = int(input("Enter a number :"))
flag = 0
for i in range(2,(num+1)//2):
    if num % i == 0:
        flag = 1
        break
if flag == 1:
    print("Not a Prime Number")
else:
    print("Prime Number")
    
```

```python
arr = [1,3,5,7,9,4,9,12,56,2]
min = arr[0]
max = arr[0]
prevMax = 0
prevMin = 0
for i in arr:
    if min > i:
        min = i
        
    if max < i:
        max = i
for i in arr:
    if prevMax > i and max > prevMax :
        prevMax = i
        
    if min < i and min < prevMin:
        prevMin = i
print("Minimum = ",min) 
print("Maximum = ",max)
print("2nd Minimum = ",prevMin) 
print("2nd Maximum = ",prevMax)
```
