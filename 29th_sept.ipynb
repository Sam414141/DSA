```python
num = int(input("Enter a number :"))
sum = 0
while num < 0:
    sum += num % 10
    num = num // 10
print(sum)
```

```python
num = int(input("Enter a number :"))
num2 = 0
tmp = num
while tmp > 0:
    num2 = (num2*10) + tmp % 10
    tmp = tmp // 10
if num2 == num:
    print(True)
else:
    print(False)
```

```python
#Armstrong
num = int(input("Enter a number :"))
l = len(str(num))
arm = 0
tmp = num
while tmp > 0:
    arm = arm + (tmp % 10)**l
    tmp = tmp // 10
if arm == num:
    print(True)
else:
    print(False)
```

```python
#HW
for i in range(1,6):
    bit = 0
    for j in range(1,10):
        if (j>=6-i and j<=4+i) and bit == 0:
            print("*",end="")
            bit = 1
        else:
            print(" ",end="")
            bit = 0
    print()
```

```python
#HW
for i in range(1,6):
    for j in range(1,10):
        if j>=6-i and j<=4+i:
            if j == 6-i | j == 4+i:
                print("*",end="")
            else:
                print(" ",end="")
        else:
            print(" ",end="")
    print()
```
