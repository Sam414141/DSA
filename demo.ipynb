```python
def count():
    for i in range(1,5):
        yield i

for i in count():
    print(i)
```

```python
def my_decorator(func):
    def wrapper():
        print("Function started")
        func()
        print("Function finished")
    return wrapper


@my_decorator
def greet():
    print("Hello!")


greet()
```

```python
def Addons(funk):
    def wrapper():
        print("Before")
        funk()
        print("After")
    return wrapper

@Addons
def fun():
    print("Having fun ,,,,")
    
fun()
```
